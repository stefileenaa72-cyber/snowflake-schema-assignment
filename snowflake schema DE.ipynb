```python
import networkx as nx
import matplotlib.pyplot as plt
```

```python
G = nx.DiGraph()
```

```python
G.add_node("Sales_Fact", color="red")
```

```python
dimensions = [
    "Customer_Dim",
    "Product_Dim",
    "Time_Dim",
    "Location_Dim"
]
for dim in dimensions:
    G.add_node(dim, color="skyblue")
    G.add_edge("Sales_Fact", dim)
```

```python
# Customer hierarchy
G.add_node("Customer_City", color="lightgreen")
G.add_node("Customer_State", color="lightgreen")
G.add_edge("Customer_Dim", "Customer_City")
G.add_edge("Customer_City", "Customer_State")
```

```python
# Product hierarchy
G.add_node("Category_Dim", color="orange")
G.add_node("Supplier_Dim", color="orange")
G.add_edge("Product_Dim", "Category_Dim")
G.add_edge("Product_Dim", "Supplier_Dim")
```

```python
G.add_node("Month_Dim", color="yellow")
G.add_node("Year_Dim", color="yellow")
G.add_edge("Time_Dim", "Month_Dim")
G.add_edge("Month_Dim", "Year_Dim")
```

```python
G.add_node("State_Dim", color="pink")
G.add_node("Country_Dim", color="pink")
G.add_edge("Location_Dim", "State_Dim")
G.add_edge("State_Dim", "Country_Dim")
```

```python
plt.figure(figsize=(14, 10))

pos = nx.spring_layout(G, seed=42)

colors = [G.nodes[node]['color'] for node in G.nodes()]

nx.draw(
    G,
    pos,
    with_labels=True,
    node_color=colors,
    node_size=3500,
    font_size=10,
    font_weight="bold",
    arrows=True
)
```
